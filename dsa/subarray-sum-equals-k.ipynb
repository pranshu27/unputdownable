```python
# ============ V0 — my brute force (kept for the record, O(n^3)) ============
# for i / for j / sum(nums[i:j+1]) inside -> the hidden third loop

# ============ V1 — prefix sums, O(n^2) ============
# sum(nums[i..j]) = pre[j+1] - pre[i]   (sentinel 0 at index 0)
def subarray_sum_v1(nums: list[int], k: int) -> int:
    n = len(nums)
    pre = [0] * (n + 1)
    for i in range(n):
        pre[i + 1] = pre[i] + nums[i]          # <- the fix: was `pre[i] += pre[i-1]`
    count = 0
    for j in range(1, n + 1):
        for i in range(j):
            if pre[j] - pre[i] == k:
                count += 1
    return count

# ============ V2 — one pass + hashmap, O(n) — THE answer ============
def subarray_sum(nums: list[int], k: int) -> int:
    """Count continuous subarrays summing to exactly k (one pass, O(n))."""
    seen = {0: 1}          # seed: empty prefix (handles subarrays starting at index 0)
    p = 0                  # running prefix sum replaces the pre[] array
    count = 0
    for x in nums:
        p += x
        count += seen.get(p - k, 0)   # lookup BEFORE record (k=0 would self-count)
        seen[p] = seen.get(p, 0) + 1
    return count
```

## Post-mortem (model answers — rewrite in your own words before moving on)
- **Brute force (V0):** enumerate O(n^2) windows; `sum()` inside makes it O(n^3) -> ~minutes at n=2e4 (TLE).
- **Optimization:** prefix sums turn a window sum into a subtraction; `pre[j]-pre[i-1]==k` iff earlier prefix `== p-k` -> hashmap lookup, one pass.
- **Why NOT sliding window:** negatives break monotonicity (growing the window can *lower* the sum). Sliding window is only valid when all nums >= 0 (contrast: LC 209 minimum-size subarray sum).
- **Complexity (V2):** O(n) time, O(n) space for the map. O(1) space is impossible with negatives in general.
- **The two classic traps:** seed `{0: 1}`; lookup-before-record when `k=0`.
- **Interview sentence:** "I progressed brute force -> prefix sums -> single-pass hashmap; the seed {0:1} counts the empty prefix so zero-start subarrays aren't missed, and the lookup must precede the insert or k=0 self-counts."
